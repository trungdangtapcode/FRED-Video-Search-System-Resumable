```python
path = '/root/data/unzipped/map-keyframes'
import os
import csv

def extract_fps_from_csv_folder(folder_path):
    fps_dict = {}

    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, 'r', newline='') as csvfile:
                reader = csv.DictReader(csvfile)
                first_row = next(reader, None)
                if first_row and 'fps' in first_row:
                    try:
                        fps_value = float(first_row['fps'])
                        name = os.path.splitext(filename)[0]
                        fps_dict[name] = fps_value
                    except ValueError:
                        pass  # Skip if fps is not a valid float

    return fps_dict
import json
save_path = "/root/data/fps_dict.json"
extract_fps_from_csv_folder(path)
with open(save_path, "w") as f:
    json.dump(extract_fps_from_csv_folder(path), f)
```
