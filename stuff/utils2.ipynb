```python
import json

def fix_truncated_json_array(path, output_path, max_lines=1000000):
    with open(path, 'r') as f:
        lines = f.readlines()

    # Try to find the last valid JSON object from the array
    valid_json_objects = []
    corrupted = False

    for line in lines:
        line = line.strip()
        if line in ['[', ']', '']:
            continue
        try:
            # Remove trailing commas and try parsing
            clean_line = line.rstrip(',')
            obj = json.loads(clean_line)
            valid_json_objects.append(obj)
        except json.JSONDecodeError:
            corrupted = True
            break

    # Save recovered data
    with open(output_path, 'w') as out:
        json.dump(valid_json_objects, out, indent=2)

    if corrupted:
        print(f"Warning: Corrupted data detected. Saved partial data to {output_path}")
    else:
        print(f"File appears to be valid. Full data saved to {output_path}")

fix_truncated_json_array(
    '/root/meilisearch/ocr/OCR/trocr+ezocr+deepsolopp+bridge+metadata.json',
    '/root/meilisearch/ocr/OCR/fixed_metadata.json'
)
```
